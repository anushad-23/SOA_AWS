```python

import boto3
import os

# Connect to DynamoDB
dynamodb = boto3.resource('dynamodb')
table = dynamodb.Table(os.environ['TABLE_NAME'])

def lambda_handler(event, context):

    action = event['action']

    # CREATE - Insert student record
    if action == 'create':

        table.put_item(
            Item={
                'StudentID': event['StudentID'],
                'Name': event['Name'],
                'Course': event['Course'],
                'Age': event['Age']
            }
        )

        return {
            'statusCode': 200,
            'body': 'Student record inserted successfully'
        }

    # READ - Retrieve student record
    elif action == 'read':

        response = table.get_item(
            Key={
                'StudentID': event['StudentID']
            }
        )

        return {
            'statusCode': 200,
            'body': response.get('Item', 'Student not found')
        }

    # UPDATE - Update student record
    elif action == 'update':

        table.update_item(
            Key={
                'StudentID': event['StudentID']
            },
            UpdateExpression='SET #n = :name, Course = :course',
            ExpressionAttributeNames={
                '#n': 'Name'
            },
            ExpressionAttributeValues={
                ':name': event['Name'],
                ':course': event['Course']
            }
        )

        return {
            'statusCode': 200,
            'body': 'Student record updated successfully'
        }

    # DELETE - Delete student record
    elif action == 'delete':

        table.delete_item(
            Key={
                'StudentID': event['StudentID']
            }
        )

        return {
            'statusCode': 200,
            'body': 'Student record deleted successfully'
        }

    else:
        return {
            'statusCode': 400,
            'body': 'Invalid action'
        }
```
